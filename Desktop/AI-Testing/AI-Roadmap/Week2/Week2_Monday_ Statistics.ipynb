Mean (μ) — the arithmetic average, Σx / n. Sensitive to outliers. Try loading the "With outlier" preset to see how one extreme value pulls the mean away from the median.

Median — the middle value of the sorted dataset. For even-length data, it's the average of the two middle values. More robust to outliers than the mean.

Variance (σ²) — average squared deviation from the mean: Σ(x − μ)² / n. Squaring serves two purposes: it makes all deviations positive, and it penalizes large deviations more heavily. This uses population variance (÷ n). For sample variance you'd use ÷ (n−1), also called Bessel's correction.
Standard deviation (σ) — just √variance, bringing units back to the original scale. Roughly, "how far a typical value is from the mean."
A few things to notice as you experiment:

When all values are equal, variance and std dev both hit 0
The mean ≠ median when data is skewed (the purple and green dashed lines on the chart show this)
Std dev grows fast when even one outlier is added — squaring the deviations amplifies extreme values

WITH SMALL INTEGERS

Step-by-step: variance

1. Values: [4, 7, 13, 2, 1]

2. Mean = 4 + 7 + 13 + 2 + 1 / 5 = 5.4

3. Squared deviations:

    (4 − 5.4)² = 1.96

    (7 − 5.4)² = 2.56

    (13 − 5.4)² = 57.76

    (2 − 5.4)² = 11.56

    (1 − 5.4)² = 19.36

4. Sum of sq. dev = 93.2

5. Variance = 93.2 / 5 = 18.64

6. Std Dev = √18.64 = 4.3174
